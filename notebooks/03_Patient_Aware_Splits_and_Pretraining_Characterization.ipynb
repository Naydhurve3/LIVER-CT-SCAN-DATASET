# 03 — Patient-Aware Splits and Pretraining Characterization

This notebook details the patient-disjoint split composition, 3D lesion morphology quartiles, and the external evaluation protocol on 3D IRCADb-01.

### Objectives:
1. Verify patient-disjoint train (104 vols), val (13 vols), and test (14 vols) splits.
2. Analyze 3D lesion morphology (845 total connected components).
3. Establish train-derived lesion stratification quartiles (Q1–Q4).
4. Define the 3D IRCADb-01 external validation protocol.

```python
import pandas as pd
import matplotlib.pyplot as plt

split_data = {
    "Split": ["Train", "Validation", "Test (Locked)"],
    "Volumes": [104, 13, 14],
    "Slices": [40667, 10685, 7286],
    "Tumor-Positive Slices": [4930, 1042, 1197],
    "Mean Tumor Burden (%)": [0.1003, 0.0874, 0.3071]
}

df_splits = pd.DataFrame(split_data)
print(df_splits.to_string(index=False))
```

## 1. 3D Lesion Size Stratification Bins (Q1 – Q4)

- **Q1 (Very Small / Small)**: $< 0.173\text{ mL}$ ($\le 6.92\text{ mm}$ equivalent diameter)
- **Q2 (Medium-Small)**: $0.173 - 0.673\text{ mL}$ ($6.92 - 10.87\text{ mm}$)
- **Q3 (Medium-Large)**: $0.673 - 3.944\text{ mL}$ ($10.87 - 19.60\text{ mm}$)
- **Q4 (Large / Massive)**: $> 3.944\text{ mL}$ ($> 19.60\text{ mm}$)

```python
quartiles = ["Q1 (Small)", "Q2 (Med-Small)", "Q3 (Med-Large)", "Q4 (Massive)"]
vol_limits = ["<0.173 mL", "0.173-0.673 mL", "0.673-3.944 mL", ">3.944 mL"]

plt.figure(figsize=(8, 4))
plt.bar(quartiles, [25, 25, 25, 25], color=["skyblue", "lightgreen", "orange", "salmon"])
plt.title("Train-Derived 3D Lesion Volume Quartile Bins", fontsize=13, fontweight="bold")
plt.ylabel("Train Lesion Population Share (%)")
for i, text in enumerate(vol_limits):
    plt.text(i, 12, text, ha="center", fontweight="bold", color="black")
plt.tight_layout()
plt.show()
```
